## Relevant Theory

### Logistic Regression

Logistic Regression is a supervised machine learning algorithm used for **binary classification**. In this experiment, it is used to classify credit card transactions into two classes:

- **0 $\rightarrow$ Legitimate transaction**
- **1 $\rightarrow$ Fraudulent transaction**

The model first calculates a linear combination of input features:

$$
z = w^T X + b
$$

where:

- $w$ = weight vector
- $X$ = input feature vector
- $b$ = bias/intercept
- $z$ = linear predictor

---

### Sigmoid Function

The Sigmoid Function converts the value of $z$ into a probability between 0 and 1:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

The predicted probability that a transaction is fraudulent is:

$$
P(y=1|X) = \sigma(w^T X + b)
$$

The predicted class can be obtained using a threshold:

$$
\hat{y} =
\begin{cases}
1, & \text{if } P(y=1|X) \geq 0.5 \\
0, & \text{if } P(y=1|X) < 0.5
\end{cases}
$$

---
## Cost Function

Logistic Regression uses the **Binary Cross-Entropy Loss** or **Log Loss**.

The cost function is:

$$
J(w,b) =
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y_i \log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

- $m$ = number of training examples
- $y_i$ = actual class label
- $\hat{y}_i$ = predicted probability

The objective of training is to find values of $w$ and $b$ that minimize the cost function.

---

## Gradient Descent

Gradient Descent is used to optimize the parameters of the Logistic Regression model.

The weights are updated using:

$$
w := w - \alpha \frac{\partial J}{\partial w}
$$

The bias is updated using:

$$
b := b - \alpha \frac{\partial J}{\partial b}
$$

where $\alpha$ is the learning rate.

For Logistic Regression, the gradient of the cost function with respect to the weights can be expressed as:

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
$$

The gradient with respect to the bias is:

$$
\frac{\partial J}{\partial b}
=
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
$$

Therefore, the parameter updates become:

$$
w := w - \alpha
\left[
\frac{1}{m}X^T(\hat{y}-y)
\right]
$$

$$
b := b - \alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
\right]
$$

---

## Class Imbalance

Credit card fraud datasets are generally highly imbalanced because legitimate transactions significantly outnumber fraudulent transactions.

For example, the dataset may contain a very large number of legitimate transactions and only a small number of fraudulent transactions.

A model that predicts most transactions as legitimate may achieve high accuracy but perform poorly at detecting fraud.

Therefore, techniques such as:

- Random Undersampling
- Oversampling
- Class Weighting

can be used to handle class imbalance.

## Evaluation Metrics

### Confusion Matrix

The confusion matrix consists of four components:

- **True Positive (TP):** Fraudulent transaction correctly identified as fraud.
- **True Negative (TN):** Legitimate transaction correctly identified as legitimate.
- **False Positive (FP):** Legitimate transaction incorrectly identified as fraud.
- **False Negative (FN):** Fraudulent transaction incorrectly identified as legitimate.

---

### Accuracy

Accuracy measures the proportion of correctly classified transactions:

$$
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
$$

However, accuracy alone is not sufficient for highly imbalanced fraud detection datasets.

---

### Precision

Precision indicates the proportion of transactions predicted as fraudulent that are actually fraudulent:

$$
Precision =
\frac{TP}
{TP+FP}
$$

A high Precision means that the model produces fewer false fraud alerts.

---

### Recall

Recall measures the proportion of actual fraudulent transactions correctly detected:

$$
Recall =
\frac{TP}
{TP+FN}
$$

Recall is particularly important in fraud detection because a **False Negative** represents a fraudulent transaction that was not detected.

---

### F1-Score

F1-Score is the harmonic mean of Precision and Recall:

$$
F1 =
2 \times
\frac{Precision \times Recall}
{Precision + Recall}
$$

---

## ROC Curve

The Receiver Operating Characteristic (ROC) curve represents the relationship between the **True Positive Rate (TPR)** and **False Positive Rate (FPR)** at different classification thresholds.

The True Positive Rate is:

$$
TPR =
\frac{TP}
{TP+FN}
$$

The False Positive Rate is:

$$
FPR =
\frac{FP}
{FP+TN}
$$

The ROC curve helps determine how well the model separates fraudulent transactions from legitimate transactions.

---

## *Area Under the Curve (AUC)*

The **Area Under the ROC Curve (AUC)** summarizes the performance of the classifier across different classification thresholds.

An AUC value close to:

$$
AUC = 1
$$

indicates excellent discrimination, while:

$$
AUC = 0.5
$$

indicates performance approximately equivalent to random classification.

Check the dataset for the following:

```python
# ============================================
# TASK 3: DATA PREPROCESSING
# Credit Card Fraud Dataset
# ============================================

# 1. Check Missing Values
print("========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal Missing Values:", missing_values.sum())


# ============================================
# 2. Handle Missing Values
# ============================================

if missing_values.sum() == 0:
    print("\nNo missing values found.")
else:
    print("\nMissing values found. Handling them using median/mode.")

    for column in df.columns:
        if df[column].isnull().sum() > 0:

            # Numerical column -> median
            if pd.api.types.is_numeric_dtype(df[column]):
                df[column] = df[column].fillna(df[column].median())

            # Categorical column -> mode
            else:
                df[column] = df[column].fillna(df[column].mode()[0])

print("\nMissing values after handling:")
print(df.isnull().sum())


# ============================================
# 3. Check Duplicate Records
# ============================================

print("\n========== DUPLICATE RECORDS ==========")

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate records removed.")
else:
    print("No duplicate records found.")

print("Dataset shape after removing duplicates:", df.shape)


# ============================================
# 4. Identify Categorical Variables
# ============================================

print("\n========== CATEGORICAL VARIABLES ==========")

categorical_columns = df.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print("Categorical Variables:")
print(categorical_columns)


# ============================================
# 5. Identify Numerical Variables
# ============================================

print("\n========== NUMERICAL VARIABLES ==========")

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical Variables:")
print(numerical_columns)


# ============================================
# 6. Separate Features and Target
# ============================================

target = 'fraud'

X = df.drop(columns=[target])
y = df[target]

print("\n========== FEATURES AND TARGET ==========")

print("Target Variable:", target)

print("\nFeature Variables:")
print(X.columns.tolist())


# ============================================
# 7. Encode Categorical Variables
# ============================================

if len(categorical_columns) > 0:

    print("\nCategorical variables found.")
    print("Applying One-Hot Encoding...")

    X = pd.get_dummies(
        X,
        columns=categorical_columns,
        drop_first=True
    )

else:

    print("\nNo categorical variables found.")
    print("Encoding is not required.")


# ============================================
# 8. Convert Boolean Columns to Integer
# ============================================

for column in X.columns:
    if X[column].dtype == 'bool':
        X[column] = X[column].astype(int)


# ============================================
# 9. Check Final Dataset
# ============================================

print("\n========== FINAL DATASET ==========")

print("Feature Dataset Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFirst Five Feature Records:")
display(X.head())

print("\nTarget Variable:")
display(y.head())


# ============================================
# 10. Check Missing Values Again
# ============================================

print("\nMissing Values in Features:")
print(X.isnull().sum().sum())

print("\nMissing Values in Target:")
print(y.isnull().sum())


# ============================================
# 11. Final Numerical Features
# ============================================

final_numerical_features = X.select_dtypes(
    include=np.number
).columns.tolist()

print("\n========== NUMERICAL FEATURES FOR LOGISTIC REGRESSION ==========")

print(final_numerical_features)

print("\nTotal Number of Numerical Features:",
      len(final_numerical_features))


# ============================================
# 12. Final Summary
# ============================================

print("\n========== TASK 3 SUMMARY ==========")

print("Missing Values Handled      : Yes")
print("Duplicate Records Removed  :", duplicate_count)
print("Categorical Variables      :", categorical_columns)
print("Number of Features         :", X.shape[1])
print("Target Variable            :", target)
print("Final Dataset Shape        :", X.shape)

print("\nDataset is ready for:")
print("1. Feature Scaling")
print("2. Train-Test Split")
print("3. Logistic Regression")
print("4. Model Evaluation")
```

## Task 4: Feature Scaling and Class Imbalance Handling

Standardize the numerical features using **Standardization** so that they have approximately zero mean and unit variance.

The standardized value of a feature is calculated as:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

* (x) = original feature value
* (\mu) = mean of the feature
* (\sigma) = standard deviation of the feature
* (z) = standardized feature value

Handle the **class imbalance** in the dataset using a suitable technique such as:

* Random Undersampling
* Oversampling
* Class Weighting

Display the class distribution **before and after** applying the selected class-imbalance handling technique.

Compare the number of legitimate and fraudulent transactions before and after balancing.

The objective of this task is to ensure that the features are on a comparable scale and that the Logistic Regression model does not become biased toward the majority class.


```python
# ============================================
# TASK 4: FEATURE SCALING AND CLASS IMBALANCE
# ============================================

from sklearn.preprocessing import StandardScaler
from sklearn.utils import resample

# ============================================
# 1. Check Class Distribution BEFORE Balancing
# ============================================

print("========== CLASS DISTRIBUTION BEFORE BALANCING ==========")

before_distribution = y.value_counts().sort_index()

print("Legitimate (0):", before_distribution.get(0, 0))
print("Fraudulent (1):", before_distribution.get(1, 0))

print("\nPercentage Distribution:")
print((y.value_counts(normalize=True) * 100).sort_index())


# ============================================
# 2. Create Dataset for Undersampling
# ============================================

# Combine features and target
data = X.copy()
data['fraud'] = y.values

# Separate majority and minority classes
majority_class = data[data['fraud'] == 0]
minority_class = data[data['fraud'] == 1]

print("\nMajority Class (Legitimate):", len(majority_class))
print("Minority Class (Fraudulent):", len(minority_class))


# ============================================
# 3. Random Undersampling
# ============================================

# Reduce legitimate transactions to the number
# of fraudulent transactions
majority_downsampled = resample(
    majority_class,
    replace=False,
    n_samples=len(minority_class),
    random_state=42
)

# Combine both classes
balanced_data = pd.concat(
    [majority_downsampled, minority_class]
)

# Shuffle the balanced dataset
balanced_data = balanced_data.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)


# ============================================
# 4. Separate Balanced Features and Target
# ============================================

X_balanced = balanced_data.drop(columns=['fraud'])
y_balanced = balanced_data['fraud']


# ============================================
# 5. Class Distribution AFTER Balancing
# ============================================

print("\n========== CLASS DISTRIBUTION AFTER BALANCING ==========")

after_distribution = y_balanced.value_counts().sort_index()

print("Legitimate (0):", after_distribution.get(0, 0))
print("Fraudulent (1):", after_distribution.get(1, 0))

print("\nPercentage Distribution:")
print(
    (y_balanced.value_counts(normalize=True) * 100).sort_index()
)


# ============================================
# 6. Compare Before and After Balancing
# ============================================

comparison = pd.DataFrame({
    'Class': ['Legitimate', 'Fraudulent'],
    'Before Balancing': [
        before_distribution.get(0, 0),
        before_distribution.get(1, 0)
    ],
    'After Balancing': [
        after_distribution.get(0, 0),
        after_distribution.get(1, 0)
    ]
})

print("\n========== BEFORE vs AFTER BALANCING ==========")
display(comparison)


# ============================================
# 7. Plot Class Distribution BEFORE Balancing
# ============================================

plt.figure(figsize=(7, 5))

plt.bar(
    ['Legitimate', 'Fraudulent'],
    [
        before_distribution.get(0, 0),
        before_distribution.get(1, 0)
    ]
)

plt.title("Class Distribution Before Balancing")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

plt.show()


# ============================================
# 8. Plot Class Distribution AFTER Balancing
# ============================================

plt.figure(figsize=(7, 5))

plt.bar(
    ['Legitimate', 'Fraudulent'],
    [
        after_distribution.get(0, 0),
        after_distribution.get(1, 0)
    ]
)

plt.title("Class Distribution After Random Undersampling")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

plt.show()


# ============================================
# 9. Feature Scaling
# ============================================

print("\n========== FEATURE SCALING ==========")

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_balanced)

# Convert back to DataFrame
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X_balanced.columns
)

print("Feature scaling completed.")

print("\nFirst 5 rows after standardization:")
display(X_scaled.head())


# ============================================
# 10. Check Mean and Standard Deviation
# ============================================

print("\n========== CHECK STANDARDIZATION ==========")

print("Mean of standardized features:")
display(X_scaled.mean().head(10))

print("\nStandard deviation of standardized features:")
display(X_scaled.std().head(10))


# ============================================
# 11. Final Dataset
# ============================================

print("\n========== FINAL DATASET ==========")

print("Final Feature Shape:", X_scaled.shape)
print("Final Target Shape :", y_balanced.shape)

print("\nFinal Class Distribution:")
print(y_balanced.value_counts())


# ============================================
# 12. Final Summary
# ============================================

print("\n========== TASK 4 SUMMARY ==========")

print("Scaling Method       : Standardization")
print("Scaling Technique    : StandardScaler")
print("Imbalance Technique  : Random Undersampling")

print("\nBefore Balancing:")
print("Legitimate :", before_distribution.get(0, 0))
print("Fraudulent :", before_distribution.get(1, 0))

print("\nAfter Balancing:")
print("Legitimate :", after_distribution.get(0, 0))
print("Fraudulent :", after_distribution.get(1, 0))

print("\nFeatures are standardized with approximately:")
print("Mean = 0")
print("Standard Deviation = 1")
```

## Task 6: Implement Sigmoid and Cost Functions

Implement the **Sigmoid Function** used in Logistic Regression.

The Sigmoid Function is defined as:

$$
\sigma(z) = \frac{1}{1+e^{-z}}
$$

The Sigmoid Function converts the output of the linear equation into a probability value between 0 and 1.

Implement the **Binary Cross-Entropy Cost Function**:

$$
J(w,b) =
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y_i\log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

* (m) = number of training samples
* (y_i) = actual class label
* (\hat{y}_i) = predicted probability
* (w) = weight vector
* (b) = bias

Test the Sigmoid Function using suitable sample values.

Calculate the initial cost of the Logistic Regression model using the initialized weights and bias.

The objective of this task is to understand how the Sigmoid Function produces probabilities and how the Binary Cross-Entropy Cost Function measures the error of the Logistic Regression model.


```python
# ============================================
# TASK 6: SIGMOID AND COST FUNCTIONS
# ============================================

import numpy as np
import matplotlib.pyplot as plt


# ============================================
# 1. Define Sigmoid Function
# ============================================

def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# ============================================
# 2. Test Sigmoid Function
# ============================================

sample_values = np.array([-5, -2, -1, 0, 1, 2, 5])

sigmoid_values = sigmoid(sample_values)

print("========== SIGMOID FUNCTION TEST ==========")

for z, value in zip(sample_values, sigmoid_values):
    print("z =", z, " -> sigmoid(z) =", round(value, 6))


# ============================================
# 3. Plot Sigmoid Function
# ============================================

z_values = np.linspace(-10, 10, 100)

plt.figure(figsize=(8, 5))
plt.plot(z_values, sigmoid(z_values))

plt.xlabel("z")
plt.ylabel("Sigmoid(z)")
plt.title("Sigmoid Function")

plt.grid(True)
plt.show()


# ============================================
# 4. Define Binary Cross-Entropy Cost Function
# ============================================

def compute_cost(X, y, w, b):

    m = len(y)

    # Linear equation
    z = np.dot(X, w) + b

    # Predicted probabilities
    y_pred = sigmoid(z)

    # Avoid log(0)
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    # Binary Cross-Entropy Cost
    cost = -(1 / m) * np.sum(
        y * np.log(y_pred) +
        (1 - y) * np.log(1 - y_pred)
    )

    return cost


# ============================================
# 5. Initialize Weights and Bias
# ============================================

# Number of features
n_features = X_train.shape[1]

# Initialize weights with zeros
w = np.zeros(n_features)

# Initialize bias with zero
b = 0


print("\n========== INITIAL PARAMETERS ==========")

print("Number of Features:", n_features)
print("Initial Weights:", w)
print("Initial Bias:", b)


# ============================================
# 6. Calculate Initial Predictions
# ============================================

z_initial = np.dot(X_train, w) + b

y_pred_initial = sigmoid(z_initial)

print("\n========== INITIAL PREDICTIONS ==========")

print("First 10 predicted probabilities:")
print(y_pred_initial[:10])


# ============================================
# 7. Calculate Initial Cost
# ============================================

initial_cost = compute_cost(
    X_train,
    y_train.values,
    w,
    b
)

print("\n========== INITIAL COST ==========")

print("Initial Cost:", initial_cost)


# ============================================
# 8. Display Actual vs Predicted Probability
# ============================================

comparison = pd.DataFrame({
    'Actual Class': y_train.values[:10],
    'Predicted Probability': y_pred_initial[:10]
})

print("\n========== ACTUAL VS PREDICTED ==========")

display(comparison)


# ============================================
# 9. Verify Sigmoid Output Range
# ============================================

print("\n========== SIGMOID OUTPUT RANGE ==========")

print("Minimum Probability:",
      y_pred_initial.min())

print("Maximum Probability:",
      y_pred_initial.max())

print("\nAll sigmoid outputs are between 0 and 1.")


# ============================================
# 10. Final Summary
# ============================================

print("\n========== TASK 6 SUMMARY ==========")

print("Sigmoid Function       : Implemented")
print("Cost Function          : Binary Cross-Entropy")
print("Initial Weights        : All zeros")
print("Initial Bias           : 0")
print("Initial Cost           :", initial_cost)
print("Training Samples       :", len(y_train))
print("Number of Features     :", n_features)
```

## Task 7: Implement Logistic Regression Using Gradient Descent

Implement **Logistic Regression from scratch** using NumPy without directly using the Scikit-learn Logistic Regression implementation.

Initialize the model parameters, including the **weight vector** (w) and **bias** (b).

Calculate the predicted probability using the Sigmoid Function:

$$
\hat{y} = \sigma(Xw+b)
$$

where:

* (X) = input feature matrix
* (w) = weight vector
* $\hat{y} = \text{predicted probability}$


Calculate the gradients of the cost function with respect to the weights and bias:

<div align="left">

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
$$

$$
\frac{\partial J}{\partial b}
=
\frac{1}{m}\sum_{i=1}^{m}(\hat{y}_i-y_i)
$$

</div>

Update the model parameters using **Gradient Descent**:

$$
w := w-\alpha\frac{\partial J}{\partial w}
$$

$$
b := b-\alpha\frac{\partial J}{\partial b}
$$

where ($\alpha$) represents the **learning rate**.

Train the model for a suitable number of iterations and record the cost value during each iteration.

Plot **Cost vs. Iteration** to verify whether the optimization process is converging.

The objective of this task is to implement Logistic Regression manually and understand how Gradient Descent optimizes the model parameters by minimizing the Binary Cross-Entropy Cost Function.


```python
# ============================================
# TASK 7: LOGISTIC REGRESSION USING
# GRADIENT DESCENT FROM SCRATCH
# ============================================

import numpy as np
import matplotlib.pyplot as plt


# ============================================
# 1. Sigmoid Function
# ============================================

def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# ============================================
# 2. Cost Function
# ============================================

def compute_cost(X, y, w, b):

    m = len(y)

    # Linear prediction
    z = np.dot(X, w) + b

    # Sigmoid prediction
    y_pred = sigmoid(z)

    # Avoid log(0)
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    # Binary Cross-Entropy Cost
    cost = -(1 / m) * np.sum(
        y * np.log(y_pred) +
        (1 - y) * np.log(1 - y_pred)
    )

    return cost


# ============================================
# 3. Gradient Descent Function
# ============================================

def gradient_descent(X, y, learning_rate=0.01, iterations=1000):

    m = X.shape[0]
    n = X.shape[1]

    # Initialize weights and bias
    w = np.zeros(n)
    b = 0

    # Store cost values
    cost_history = []

    # Gradient Descent
    for i in range(iterations):

        # ----------------------------
        # Forward Propagation
        # ----------------------------

        z = np.dot(X, w) + b

        y_pred = sigmoid(z)

        # ----------------------------
        # Calculate Gradients
        # ----------------------------

        dw = (1 / m) * np.dot(X.T, (y_pred - y))

        db = (1 / m) * np.sum(y_pred - y)

        # ----------------------------
        # Update Parameters
        # ----------------------------

        w = w - learning_rate * dw

        b = b - learning_rate * db

        # ----------------------------
        # Calculate Cost
        # ----------------------------

        cost = compute_cost(X, y, w, b)

        cost_history.append(cost)

        # Display cost every 100 iterations
        if (i + 1) % 100 == 0:
            print(
                "Iteration:",
                i + 1,
                "Cost:",
                round(cost, 6)
            )

    return w, b, cost_history


# ============================================
# 4. Prepare Training Data
# ============================================

X_train_np = np.asarray(X_train, dtype=float)
y_train_np = np.asarray(y_train, dtype=float)

print("Training Data Shape:", X_train_np.shape)
print("Training Target Shape:", y_train_np.shape)


# ============================================
# 5. Set Hyperparameters
# ============================================

learning_rate = 0.01
iterations = 1000

print("\n========== HYPERPARAMETERS ==========")
print("Learning Rate:", learning_rate)
print("Iterations:", iterations)


# ============================================
# 6. Train Logistic Regression Model
# ============================================

print("\n========== TRAINING MODEL ==========")

w_final, b_final, cost_history = gradient_descent(
    X_train_np,
    y_train_np,
    learning_rate=learning_rate,
    iterations=iterations
)


# ============================================
# 7. Display Final Parameters
# ============================================

print("\n========== FINAL PARAMETERS ==========")

print("Final Weights:")
print(w_final)

print("\nFinal Bias:")
print(b_final)

print("\nInitial Cost:")
print(cost_history[0])

print("\nFinal Cost:")
print(cost_history[-1])


# ============================================
# 8. Plot Cost vs Iteration
# ============================================

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, iterations + 1),
    cost_history
)

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration - Gradient Descent")

plt.grid(True)
plt.show()


# ============================================
# 9. Check Cost Reduction
# ============================================

print("\n========== COST REDUCTION ==========")

print("Cost at Iteration 1   :", cost_history[0])
print("Cost at Iteration 100 :", cost_history[99])
print("Cost at Iteration 500 :", cost_history[499])
print("Cost at Iteration 1000:", cost_history[-1])

if cost_history[-1] < cost_history[0]:
    print("\nCost decreased successfully.")
    print("Gradient Descent is converging.")
else:
    print("\nCost did not decrease.")
    print("Try changing the learning rate.")


# ============================================
# 10. Function for Prediction
# ============================================

def predict_probability(X, w, b):

    z = np.dot(X, w) + b

    return sigmoid(z)


def predict_class(X, w, b, threshold=0.5):

    probabilities = predict_probability(X, w, b)

    return (probabilities >= threshold).astype(int)


# ============================================
# 11. Test Predictions on Training Data
# ============================================

training_probabilities = predict_probability(
    X_train_np,
    w_final,
    b_final
)

training_predictions = predict_class(
    X_train_np,
    w_final,
    b_final
)

print("\n========== SAMPLE PREDICTIONS ==========")

for i in range(10):
    print(
        "Actual:",
        int(y_train_np[i]),
        "| Probability:",
        round(training_probabilities[i], 4),
        "| Predicted:",
        training_predictions[i]
    )


# ============================================
# 12. Final Summary
# ============================================

print("\n========== TASK 7 SUMMARY ==========")

print("Algorithm              : Logistic Regression")
print("Implementation         : From Scratch")
print("Library Used           : NumPy")
print("Learning Method        : Gradient Descent")
print("Learning Rate          :", learning_rate)
print("Number of Iterations   :", iterations)
print("Initial Cost           :", cost_history[0])
print("Final Cost             :", cost_history[-1])

print("\nModel training completed successfully.")
```

## Task 9: Calculate Precision, Recall and F1-Score

Calculate the following performance metrics for the Logistic Regression model:

### Precision

Precision measures the proportion of transactions predicted as fraudulent that are actually fraudulent.

$$
\text{Precision} =
\frac{TP}{TP+FP}
$$

### Recall

Recall measures the proportion of actual fraudulent transactions that are correctly identified by the model.

$$
\text{Recall} =
\frac{TP}{TP+FN}
$$

### F1-Score

The F1-Score is the harmonic mean of Precision and Recall.

$$
F1\text{-Score} =
2 \times
\frac{\text{Precision} \times \text{Recall}}
{\text{Precision}+\text{Recall}}
$$

### Accuracy

Calculate accuracy for comparison:

$$
\text{Accuracy} =
\frac{TP+TN}
{TP+TN+FP+FN}
$$

Display the calculated **Accuracy, Precision, Recall, and F1-Score**.

Therefore, a model with high accuracy but low Recall may fail to detect a significant number of fraudulent transactions.

The objective of this task is to evaluate the Logistic Regression model using appropriate classification metrics and understand the importance of Recall in credit card fraud detection.


```python
# Task 9: Calculate Precision, Recall, F1-Score and Accuracy

# Calculate Accuracy
accuracy = (TP_test + TN_test) / (TP_test + TN_test + FP_test + FN_test)

# Calculate Precision
precision = TP_test / (TP_test + FP_test) if (TP_test + FP_test) != 0 else 0

# Calculate Recall
recall = TP_test / (TP_test + FN_test) if (TP_test + FN_test) != 0 else 0

# Calculate F1-Score
f1_score = (2 * precision * recall) / (precision + recall) if (precision + recall) != 0 else 0

# Display results
print("===== LOGISTIC REGRESSION PERFORMANCE =====")
print(f"Accuracy  : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1_score:.4f}")
```
